```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

# Sample data
np.random.seed(42)

# Simplest pairplot — full grid
sns.pairplot(df)
plt.suptitle('Pair Plot (Full Grid)', y=1.02)
plt.show()
```

```python


# Select only numeric columns
numeric_df = df.select_dtypes(include=[np.number])

# Compute correlation matrix
corr_matrix = numeric_df.corr()

# Plot heatmap
plt.figure(figsize=(10, 8))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt='.2f',
            vmin=-1, vmax=1, center=0, square=True, linewidths=0.5)
plt.title('Correlation Matrix Heatmap')
plt.tight_layout()
plt.show()
```

```python
for col in df.columns:
    sns.countplot(df, x=col, hue="target")
    plt.title(col)
    plt.show()
```

```python


grouped = df.drop(columns="age").groupby('target').mean(numeric_only=True)

plt.figure(figsize=(10, 6))
sns.heatmap(grouped, annot=True, fmt='.2f', cmap='coolwarm', cbar=True)
plt.title('Mean values by target')
plt.ylabel('target')
plt.xlabel('features')
plt.tight_layout()
plt.show()
```
